## Lấy ra danh sách employee (Người quản lý invoice tạo, không phải là bác sĩ thú y)

```python
conn.rollback()
query = """
  SELECT e.employee_id
  FROM employee e
  LEFT JOIN veterinarian v 
    ON e.employee_id = v.employee_id
  WHERE v.employee_id IS NULL;
"""
cur.execute(query)
employees = cur.fetchall()
employees = [emp for (emp,) in employees]
for emp in employees:
    print(emp)
print(len(employees))
```

## Lấy ra danh sách khách hàng và thú cưng của họ

```python
query = """
  SELECT customer.customer_id, pet.pet_id
  FROM customer
  JOIN pet ON customer.customer_id = pet.customer_id
"""

cur.execute(query)
rows = cur.fetchall()

customers_map = {}

for customer_id, pet_id in rows:
    if customer_id not in customers_map:
        customers_map[customer_id] = {
            "id": customer_id,
            "pets": []
        }
    customers_map[customer_id]["pets"].append(pet_id)

customers = list(customers_map.values())

print(len(customers))
for customer in customers:
    print(customer["id"])
    print(customer["pets"])
```

## Lấy danh sách Product

```python
conn.rollback()
query = """
  SELECT product.product_id, product.price
  FROM product
"""

cur.execute(query)
products = cur.fetchall()

def format_product(row):
    return {
        "id": row[0],
        "price": float(row[1]),
    }

products = [format_product(row) for row in products]

print(len(products))
for accessory in products:
    print(accessory)
```

## Main

```python
# Mỗi KH có nhiều invoice
conn.rollback()
for customer in customers:
  # Tạo mới invoices cho mỗi KH

  def random_datetime(start_dt, end_dt):
    delta = end_dt - start_dt
    random_seconds = random.randint(0, int(delta.total_seconds()))
    return start_dt + datetime.timedelta(seconds=random_seconds)

  start_dt = datetime.datetime(2022, 1, 1, 0, 0, 0)
  end_dt = datetime.datetime.now()

  new_invoices = {
    "created_at": random_datetime(start_dt, end_dt),
    "payment_method": random.choice(["cash", "credit_card", "momo"]),
    "total_price": 0,
    "employee_id": random.choice(employees),
    "customer_id": customer["id"],
  }
  query = """
    INSERT INTO invoice (created_at, payment_method, total_price, employee_id, customer_id)
    VALUES (%(created_at)s, %(payment_method)s, %(total_price)s, %(employee_id)s, %(customer_id)s)
    RETURNING invoice_id;
  """
  cur.execute(query, new_invoices)
  invoice_id = cur.fetchone()[0]
  
  # Tạo chi tiết hóa đơn product
  num_products = random.randint(1, 5)
  selected_products = random.sample(products, num_products)
  invoiceproducts = []
  for product in selected_products:
    invoiceproducts.append({
      "invoice_id": invoice_id,
      "product_id": product["id"],
      "quantity": random.randint(1, 3),
    })
  query = """
    INSERT INTO invoiceproduct (invoice_id, product_id, quantity)
    VALUES %s;
  """
  execute_values(cur, query, [(ip["invoice_id"], ip["product_id"], ip["quantity"]) for ip in invoiceproducts])
  conn.commit()
```
